1. What is Inferential Statistics?
Inferential statistics is a method of using a small sample of data to understand a large group of people.
It helps us make predictions and take decisions based on data.

2. What is Hypothesis Testing and Its Components?
Hypothesis testing is used to check whether a statement is true or false with the help of data. 
Its main components are null hypothesis, alternative hypothesis, significance level, test statistic, p-value, and conclusion.

3. Explain Confidence Interval and Critical Value.
A confidence interval is a range in which the actual value of a population is expected to lie.
 A critical value is a limit used to decide whether we should reject the null hypothesis.

4. Define P-value.
A p-value helps us understand whether the result of a statistical test is significant or not. 
If the p-value is less than 0.05, we generally reject the null hypothesis.

5. Differentiate Type I and Type II Errors.
Type I Error: It happens when we reject the null hypothesis even though it is true.
Type II Error: It happens when we do not reject the null hypothesis even though it is false.

6. Brief Descriptions of Z-test, T-test, Chi-square Test, and ANOVA Test.
Z-test: It is used to compare a sample mean or proportion with a known value when the conditions for the test are satisfied.
T-test: It is used to compare the average values of two groups.
Chi-square Test: It is used to check the relationship between two categorical variables, such as smoking and diabetes.
ANOVA Test: It is used to compare the average values of three or more groups.

7. What is Covariance?
Covariance tells us how two variables change with each other. If both variables tend to increase or decrease together, covariance is positive.
If one increases while the other decreases, it is negative.

8. What is Correlation?
Correlation tells us how strongly two variables are related. Its value ranges from -1 to +1.
Positive correlation means both variables move in the same direction, while negative correlation means they move in opposite directions.
